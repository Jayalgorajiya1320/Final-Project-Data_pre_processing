## Q6. Handle Missing Data with Different Imputation Techniques

```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Users/JAYAL/Downloads/customer_credit_risk_dataset.csv")

print(df.isnull().sum())
```

## 1. Simple Imputer (Numerical: Mean / Median)

## A. Mean Imputation

```python
from sklearn.impute import SimpleImputer

num_cols = ['age', 'annual_income', 'credit_score']

df_mean = df.copy()

mean_imputer = SimpleImputer(strategy='mean')

df_mean[num_cols] = mean_imputer.fit_transform(
    df_mean[num_cols]
)

print(df_mean[num_cols].isnull().sum())
```

## B. Median Imputation

```python
df_median = df.copy()

median_imputer = SimpleImputer(strategy='median')

df_median[num_cols] = median_imputer.fit_transform(
    df_median[num_cols]
)

print(df_median[num_cols].isnull().sum())
```

## 2. Simple Imputer (Categorical: Most Frequent)

```python
cat_cols = ['gender', 'employment_type']

df_cat = df.copy()

cat_imputer = SimpleImputer(strategy='most_frequent')

df_cat[cat_cols] = cat_imputer.fit_transform(
    df_cat[cat_cols]
)

print(df_cat[cat_cols].isnull().sum())
```

## 3. Most Frequent Category Imputation

```python
df_mode = df.copy()

for col in ['gender', 'employment_type']:
    mode_value = df_mode[col].mode()[0]
    df_mode[col] = df_mode[col].fillna(mode_value)

print(df_mode[['gender', 'employment_type']].isnull().sum())
```

## 4. Missing Indicator + Random Sample Imputation

```python
df_random = df.copy()

col = 'annual_income'

df_random[col + '_missing'] = (
    df_random[col].isnull().astype(int)
)

# Random sample imputation
missing = df_random[col].isnull()

df_random.loc[missing, col] = np.random.choice(
    df_random[col].dropna(),
    size=missing.sum(),
    replace=True
)

print(df_random[[col, col + '_missing']].head())
```

## 5. KNN Imputer (Multivariate)

```python
from sklearn.impute import KNNImputer

df_knn = df.copy()

num_cols = ['age', 'annual_income', 'loan_amount',
            'credit_score', 'spending_ratio']

knn_imputer = KNNImputer(n_neighbors=5)

df_knn[num_cols] = knn_imputer.fit_transform(
    df_knn[num_cols]
)

print(df_knn[num_cols].isnull().sum())
```

## 6. MICE Algorithm (Multivariate Imputation)

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df_mice = df.copy()

num_cols = ['age', 'annual_income', 'credit_score']

mice_imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

df_mice[num_cols] = mice_imputer.fit_transform(
    df_mice[num_cols]
)

print(df_mice[num_cols].isnull().sum())
```

## Part D: Outlier Handling

## Q7. Detect and Treat Outliers Using Different Methods

## 1. Z-Score Method

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

df = pd.read_csv("C:/Users/JAYAL/Downloads/customer_credit_risk_dataset.csv")

# Select numerical column
col = 'annual_income'

# Calculate Z-score
z_scores = zscore(df[col], nan_policy='omit')

# Detect outliers
outliers = df[np.abs(z_scores) > 3]

print("Outliers:")
print(outliers[['customer_id', col]])

print("Total Outliers:", len(outliers))
```

```python
df_zscore = df.copy()

z = np.abs(z_scores)

df_zscore = df_zscore.loc[z <= 3].copy()

print(df_zscore.shape)
```

## 2. IQR Method (Interquartile Range)

```python
col = 'annual_income'

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df[col] < lower) |
    (df[col] > upper)
]

print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Outliers:")
print(outliers[['customer_id', col]])
```

```python
df_iqr = df.copy()

df_iqr[col] = df_iqr[col].clip(
    lower=lower,
    upper=upper
)

print(df_iqr[col].describe())
```

## 3. Percentile Method

```python
col = 'loan_amount'

lower = df[col].quantile(0.05)
upper = df[col].quantile(0.95)

outliers = df[
    (df[col] < lower) |
    (df[col] > upper)
]

print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Total Outliers:", len(outliers))
```

```python
df_percentile = df.copy()

df_percentile[col] = df_percentile[col].clip(
    lower=lower,
    upper=upper
)

print(df_percentile[col].describe())
```

## 4. Winsorization Technique

```python
from scipy.stats.mstats import winsorize

df_winsor = df.copy()

col = 'annual_income'

df_winsor[col] = winsorize(
    df_winsor[col],
    limits=[0.05, 0.05]
)

print(df_winsor[col].describe())
```

## Q8. Handle Variable Types

## 1. Mixed Variables (Numerical + Categorical)

```python
import pandas as pd

df = pd.read_csv("C:/Users/JAYAL/Downloads/customer_credit_risk_dataset.csv")

# Select numerical columns
num_cols = df.select_dtypes(
    include=['int64', 'float64']
).columns

# Select categorical columns
cat_cols = df.select_dtypes(
    include=['object']
).columns

print("Numerical Columns:")
print(num_cols)

print("\nCategorical Columns:")
print(cat_cols)
```

## 2. Date & Time Variables (Extract Year, Month, Day, Weekday)

```python
df['join_date'] = pd.to_datetime(
    df['join_date']
)

# Extract Year
df['year'] = df['join_date'].dt.year

# Extract Month
df['month'] = df['join_date'].dt.month

# Extract Day
df['day'] = df['join_date'].dt.day

# Extract Weekday
df['weekday'] = df['join_date'].dt.day_name()

print(df[['join_date', 'year',
          'month', 'day', 'weekday']].head())
```

## 2. Label Encoding (Gender)

```python
from sklearn.preprocessing import LabelEncoder

df['gender'] = df['gender'].fillna(
    df['gender'].mode()[0]
)

le = LabelEncoder()

df['gender_encoded'] = le.fit_transform(
    df['gender']
)

print(df[['gender', 'gender_encoded']].head())
```

## 3. One-Hot Encoding (Region and Loan Purpose)

```python
df_encoded = pd.get_dummies(
    df,
    columns=['region', 'loan_purpose'],
    dtype=int
)

print(df_encoded.head())
```

## Q10. Encoding Numerical Features

## 1. Binning (Discretize Income into Groups)

```python
df['income_group'] = pd.cut(
    df['annual_income'],
    bins=[0, 300000, 700000,
          1200000, float('inf')],
    labels=['Low', 'Medium',
            'High', 'Very High']
)

print(df[['annual_income',
          'income_group']].head())
```

## 2. Binarization (Flag if Credit Score > Threshold)

```python
df['high_credit_flag'] = np.where(
    df['credit_score'] > 700,
    1,
    0
)

print(df[['credit_score',
          'high_credit_flag']].head())
```

## 3. Quantile Binning

```python
df['transaction_group'] = pd.qcut(
    df['transaction_count'],
    q=4,
    labels=['Low', 'Medium',
            'High', 'Very High'],
    duplicates='drop'
)

print(df[['transaction_count',
          'transaction_group']].head())
```

## 4. K-Means Binning

```python
from sklearn.cluster import KMeans

# Prepare data
X = df[['transaction_count']].dropna()

# Create KMeans model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Fit model and predict clusters
clusters = kmeans.fit_predict(X)

# Store results
df['transaction_cluster'] = np.nan

df.loc[X.index, 'transaction_cluster'] = clusters

print(df[['transaction_count',
          'transaction_cluster']].head())
```

## Q12. Apply Transformations

## 3. ColumnTransformer

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.read_csv("C:/Users/JAYAL/Downloads/customer_credit_risk_dataset.csv")

num_cols = [
    'age',
    'annual_income',
    'loan_amount',
    'credit_score'
]

cat_cols = [
    'gender',
    'region',
    'employment_type',
    'loan_purpose'
]

# Numerical preprocessing
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

# Apply transformations
X = preprocessor.fit_transform(df)

print(X.shape)
print("Transformation completed!")
```

## Part H: Final Deliverable

## Q14. Provide a Final Cleaned and Transformed Dataset

## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

## Step 2: Load Dataset

```python
df = pd.read_csv("customer_credit_risk_dataset.csv")

print("Original Shape:", df.shape)

# Remove duplicate records
df = df.drop_duplicates()
```

## Step 3: Construct New Features

```python
df['debt_to_income_ratio'] = (
    df['loan_amount'] / df['annual_income']
)

df['avg_monthly_transactions'] = (
    df['transaction_count'] / 6
)

df['debt_to_income_ratio'] = (
    df['debt_to_income_ratio'].replace(
        [np.inf, -np.inf], np.nan
    )
)
```

## Step 4: Select Numerical and Categorical Columns

```python
num_cols = [
    'age',
    'annual_income',
    'loan_amount',
    'credit_score',
    'repayment_history',
    'transaction_count',
    'spending_ratio',
    'debt_to_income_ratio',
    'avg_monthly_transactions'
]

cat_cols = [
    'gender',
    'region',
    'education_level',
    'employment_type',
    'loan_purpose'
]
```

## Step 5: Create Preprocessing Pipelines

```python
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(
        strategy='most_frequent'
    )),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore'
    ))
])
```

## Step 6: Apply ColumnTransformer

```python
preprocessor = ColumnTransformer([
    ('numerical', num_pipeline, num_cols),
    ('categorical', cat_pipeline, cat_cols)
])

X = preprocessor.fit_transform(df)

print("Preprocessing Completed!")
print("Transformed Shape:", X.shape)
```

## Step 7: Create Final Dataset

```python
feature_names = preprocessor.get_feature_names_out()

final_df = pd.DataFrame(
    X,
    columns=feature_names,
    index=df.index
)

final_df.insert(
    0, 'customer_id', df['customer_id']
)

final_df['default_flag'] = df['default_flag']

print(final_df.head())
print(final_df.shape)
```

## Step 8: Save Final Dataset

```python
final_df.to_csv(
    "final_cleaned_transformed_dataset.csv",
    index=False
)

print("Final Dataset Saved Successfully!")
```
